```python
import os
import torch
from torchvision import datasets, transforms
from torch.utils.data import DataLoader

# --- Calculate Dataset RGB Mean and Standard Deviation ---
dataset_path = "../dataset"

# Define transform to convert images to tensors
transform = transforms.Compose([
    transforms.Resize((224, 224)),  # Standard model input size check
    transforms.ToTensor()
])

if os.path.exists(dataset_path):
    print("Loading dataset for normalization statistics calculation...")
    dataset = datasets.ImageFolder(root=dataset_path, transform=transform)
    dataloader = DataLoader(dataset, batch_size=64, shuffle=False, num_workers=4)

    mean = torch.zeros(3)
    std = torch.zeros(3)
    total_images = len(dataset)

    print(f"Processing {total_images} images across batches...")
    for images, _ in dataloader:
        # Batch shape: [batch_size, channels, height, width]
        batch_samples = images.size(0)
        images = images.view(batch_samples, images.size(1), -1)

        # Sum mean and std over batch pixels
        mean += images.mean(2).sum(0)
        std += images.std(2).sum(0)

    mean /= total_images
    std /= total_images

    print("\n" + "="*45)
    print("DATASET NORMALIZATION STATISTICS")
    print("="*45)
    print(f"Calculated RGB Mean : {mean.tolist()}")
    print(f"Calculated RGB Std  : {std.tolist()}")
    print("="*45)
else:
    print(f"Error: {dataset_path} does not exist")
```
