```python
import os
import cv2
import numpy as np
import pandas as pd

# --- Brightness Analysis ---
dataset_path = "../dataset"
brightness_data = []

# Thresholds for extreme dark and bright images (0-255 scale)
dark_threshold = 30.0
bright_threshold = 225.0

if os.path.exists(dataset_path):
    print("Starting brightness analysis...")
    for class_name in sorted(os.listdir(dataset_path)):
        class_dir = os.path.join(dataset_path, class_name)
        if os.path.isdir(class_dir):
            for f in os.listdir(class_dir):
                if f.lower().endswith((".jpg", ".jpeg", ".png")):
                    file_path = os.path.join(class_dir, f)
                    try:
                        # Read image in grayscale using OpenCV
                        img = cv2.imread(file_path, cv2.IMREAD_GRAYSCALE)
                        if img is not None:
                            # Calculate average pixel intensity (brightness score)
                            mean_brightness = np.mean(img)
                            brightness_data.append({
                                "Class": class_name,
                                "File_Path": file_path,
                                "Brightness": mean_brightness
                            })
                        else:
                            print(f"Could not read image: {file_path}")
                    except Exception as e:
                        print(f"Error processing file {file_path}: {e}")

    df_brightness = pd.DataFrame(brightness_data)
    print(f"Total scanned images for brightness analysis: {len(df_brightness)}")
else:
    print(f"Error: {dataset_path} does not exist")
```

```python
# --- Analyze Brightness Distribution and Outliers ---
if 'df_brightness' in locals() and not df_brightness.empty:
    print("\n--- Brightness Score Statistics ---")
    print(df_brightness["Brightness"].describe())

    # Filter extreme dark and bright images
    dark_images = df_brightness[df_brightness["Brightness"] < dark_threshold]
    bright_images = df_brightness[df_brightness["Brightness"] > bright_threshold]

    print(f"\nNumber of extremely dark images (brightness < {dark_threshold}): {len(dark_images)}")
    print(f"Number of extremely bright images (brightness > {bright_threshold}): {len(bright_images)}")

    # Report summary
    print("\n" + "="*45)
    print("BRIGHTNESS ANALYSIS REPORT")
    print("="*45)
    print(f"Total Images Analyzed    : {len(df_brightness)}")
    print(f"Extremely Dark Images    : {len(dark_images)}")
    print(f"Extremely Bright Images  : {len(bright_images)}")
    print(f"Total Outliers (Lighting): {len(dark_images) + len(bright_images)}")
    print("="*45)
```
