```python
# Imports
import os
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns
```

```python
# --- 1. Distribution Analysis ---

# Dataset Path
dataset_path = "../dataset"

# Dictionary to keep classes and images
class_counts = {}

# Folder Iteration
if os.path.exists(dataset_path):
    for class_name in sorted(os.listdir(dataset_path)):
        class_dir = os.path.join(dataset_path, class_name)

        # Examine folder ones
        if os.path.isdir(class_dir):
            # Counting .jpg extensions
            images = [
                f
                for f in os.listdir(class_dir)
                if f.lower().endswith((".jpg"))
            ]
            class_counts[class_name] = len(images)
    # Converting to Pandas DataFrame
    df_balance = pd.DataFrame(list(class_counts.items()), columns=["Class", "Count"])
    df_balance = df_balance.sort_values(by="Count", ascending=False).reset_index(drop=True)
else:
    print(f"Error: {dataset_path} does not exist")
```

```python
# --- 2. Visualization ---
# Showing the table
print("--- Class Distribution ---")
display(df_balance)

# Virtualization (Bar Plot)
plt.figure(figsize=[14, 6])
sns.barplot(
    data=df_balance,
    x="Class",
    y="Count",
    hue="Class",
    palette="Blues_r",
    legend=False,
)
plt.xticks(rotation=90)
plt.title("Turkish Sign Language - Class Balance", fontsize=14)
plt.xlabel("Classes", fontsize=12)
plt.ylabel("Count", fontsize=12)
plt.grid(axis="y",linestyle="--",alpha=0.7)
plt.tight_layout()
plt.show()
```
